# Olist E-commerce Data Exploration
This notebook explores the basic metrics and relationships of the Olist dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
import os
sys.path.append(os.path.abspath('..'))
from data_loader import load_all_data

data = load_all_data()
orders = data['orders']
reviews = data['reviews']
products = data['products']
```

## 1. Data Shapes

```python
for name, df in data.items():
    print(f"{name}: {df.shape}")
```

## 2. Null Values

```python
for name, df in data.items():
    print(f"\n--- {name.upper()} Null Values ---")
    nulls = df.isnull().sum()
    print(nulls[nulls > 0])
```

## 3. Data Analysis

```python
# Top Order Statuses
print("Top Order Statuses:")
order_status_counts = orders['order_status'].value_counts()
print(order_status_counts)
order_status_counts.plot(kind='bar', title='Order Status Distribution', figsize=(8,4))
plt.show()
```

```python
# Average Review Score
avg_score = reviews['review_score'].mean()
print(f"Average Review Score: {avg_score:.2f}")
reviews['review_score'].value_counts().sort_index().plot(kind='bar', title='Review Score Distribution', color='orange', figsize=(8,4))
plt.show()
```

```python
# Top Product Categories
print("Top 5 Product Categories:")
top_categories = products['product_category_name'].value_counts().head(5)
print(top_categories)
top_categories.plot(kind='bar', title='Top Product Categories', color='green', figsize=(8,4))
plt.show()
```

```python
# Delayed Deliveries Count
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
delayed_orders = orders[orders['order_delivered_customer_date'] > orders['order_estimated_delivery_date']]
print(f"Total Delayed Deliveries: {len(delayed_orders)}")
```
